```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.lineplot(data=gold_25, x="year",y="price")
plt.title("Tendencia Oro")
plt.xlabel("Año")
plt.ylabel("Precio")
```

```python
#Precio de cierre de cada año
precio_anual = gold_25.groupby("year")["price"].last()
precio_anual
```

```python
#Porcentaje de crecimiento respecto al año anterior
crecimiento = precio_anual.pct_change() * 100
crecimiento
```

```python
# Rendimiento promedio anual
crecimiento.mean()
```

```python
mejor_año = crecimiento.idxmax()
mejor_año
```

```python
mejor_valor = crecimiento.max()
mejor_valor
```

```python
peor_año = crecimiento.idxmin()
peor_año
```

```python
peor_valor = crecimiento.min()
peor_valor
```

```python
print(f"El mejor año fue:",mejor_año,"y el porcentaje de variacion fue de:", mejor_valor)
print(f"El peor año fue:",peor_año,"y el porcentaje de variacion fue de:", peor_valor)
```

```python
# Volatilidad del precio del oro
vol_mensual = gold_25["pct_change_1"].std()
vol_anual = vol_mensual * np.sqrt(12)
vol_anual
```
